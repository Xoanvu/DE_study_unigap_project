x before swapping: 5

## Even Or Odd

Given an integer, print whether it is Even or Odd.

----------------------------------------------------------------------
#### Input:
An integer

#### Output:
'Even' or 'Odd'

----------------------------------------------------------------------
#### Sample input:
11

#### Sample output:
Odd

----------------------------------------------------------------------
#### Sample input:
2

#### Sample output:
Even

````python
def check(prompt="Enter an integer: "):
    while True:
        try:
            x = int(input(prompt))
            return x
        except ValueError:
            print('The input value is not an integer! Please try again.')

x = check()
if x%2 == 0:
    print("Even")
else:
    print("Odd")
````

## Alarm Clock

You're trying to automate your alarm clock by writing a function for it. You're given a day of the week encoded as 1=Mon, 2=Tue, ... 6=Sat, 7=Sun, and whether you are on vacation as a boolean value (a boolean object is either True or False. Google "booleans python" to get a better understanding). Based on the day and whether you're on vacation, write a function that returns a time in form of a string indicating when the alarm clock should ring. 

When not on a vacation, on weekdays, the alarm should ring at "7:00" and on the weekends (Saturday and Sunday) it should ring at "10:00". 

While on a vacation, it should ring at "10:00" on weekdays. On vacation, it should not ring on weekends, that is, it should return "off".

----------------------------------------------------------------------
#### Input:
The input will be a list of two elements. The first element will be an integer from 1 to 7, and the second element will be a boolean value.

#### Output:
The output will be a string denoting the time alarm will ring or 'off'

----------------------------------------------------------------------
#### Sample input:
[7, True]

#### Sample output:
off

----------------------------------------------------------------------
#### Sample input:
[3, True]

#### Sample output:
10:00

----------------------------------------------------------------------

````python
import ast

def validate_input(inp: str):
    try:
        arr = ast.literal_eval(inp)

        if not isinstance(arr, list):
            raise ValueError("Input must be a list [int, bool]")
        
        if len(arr) != 2:
            raise ValueError("Input must have exactly 2 elements [int, bool]")
        
        day_of_week, is_vacation = arr

        if not isinstance(day_of_week, int) or not (1 <= day_of_week <= 7):
            raise ValueError("First element must be an integer between 1 and 7")

        if not isinstance(is_vacation, bool):
            raise ValueError("Second element must be a boolean (True/False)")

        return day_of_week, is_vacation

    except Exception as e:
        raise ValueError(f"Invalid input format: {e}")


def set_alarm_clock(day_of_week: int, is_vacation: bool) -> str:
    if not isinstance(day_of_week, int) or not (1 <= day_of_week <= 7):
        raise ValueError("day_of_week must be an integer between 1 and 7")

    if not isinstance(is_vacation, bool):
        raise ValueError("is_vacation must be a boolean (True/False)")
    
    if is_vacation:
        if day_of_week in (6, 7):
            return "off"
        else:
            return "10:00"
    else:
        if day_of_week in (6, 7):
            return "10:00"
        else:
            return "7:00"
        
inp = input("Enter values to calculate alarm time (format: [day_of_week 1-7, is_vacation True/False]): ")

while True:
    try:
        day_of_week, is_vacation = validate_input(inp)
        alarm_clock = set_alarm_clock(day_of_week, is_vacation)
        print(alarm_clock)
        break
    except ValueError as e:
        print(e)
````

## Factorial

Factorial is a mathematical function denoted by '!'. It is defined as

n factorial = n!= 1*2*3...*(n-1)*n

In this question, you have to make a function that will take an integer as input, and return the factorial of that integer if that integer is greater than or equal to zero and return -1 if the number is less than zero or negative.

Note: the function doesn't return print the factorial but returns it.

----------------------------------------------------------------------
#### Input:
An integer n

#### Output:
The function returns n! if n is greater than or equal to 0.
0! = 1
and the function returns -1 if the number is less than 0.

----------------------------------------------------------------------
#### Sample input:
3

#### Sample output:
6

----------------------------------------------------------------------
#### Sample input:
-4

#### Sample output:
-1

````python
def check(prompt="Enter an integer: "):
    while True:
        try:
            n = int(input(prompt))
            return n
        except ValueError:
            print('The input value is not an integer! Please try again.')

n = check()
result = 1
if n < 0:
    print(-1)
elif n == 0:
    print(1)
else:
    for i in range(1, n + 1):
        result = result * i

print(result)


````

## Reverse The Digits

You will be given a number. You have to reverse the digits of the number and print it.

----------------------------------------------------------------------
#### Input:
A positive integer greater than zero

#### Output:
The number in reverse order. Check sample outputs for more details.

----------------------------------------------------------------------
#### Sample input:
345200

#### Sample output:
2543

----------------------------------------------------------------------
#### Sample input:
6752343

#### Sample output:
3432576

````python
def check(prompt="Enter an integer: "):
    while True:
        try:
            n = int(input(prompt))
            n > 0
            return n
        except ValueError:
            print('The input value is not an integer! Please try again.')

n = check()
reverse = int(str(n)[::-1])
print(reverse)
````
